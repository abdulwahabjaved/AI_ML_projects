```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
df = sns.load_dataset('iris')

print("Dataset loaded successfully!")
print(type(df))
```

```python
print("Shape:", df.shape)
print("\nColumn Names:", df.columns.tolist())
```

```python
df.info()
```

```python
df.describe()
```

```python
print("Unique Species:", df['species'].unique())
print("\nSpecies Count:")
print(df['species'].value_counts())
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='sepal_length', y='sepal_width', hue='species', palette='Set1', s=80)
plt.title('Sepal Length vs Sepal Width')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')
plt.legend(title='Species')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='petal_length', y='petal_width', hue='species', palette='Set2', s=80)
plt.title('Petal Length vs Petal Width')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Petal Width (cm)')
plt.legend(title='Species')
plt.tight_layout()
plt.show()
```

```python
sns.pairplot(df, hue='species', palette='husl', height=2.5)
plt.suptitle('Pairplot of All Features', y=1.02)
plt.show()
```

```python
features = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
fig.suptitle('Histograms: Feature Distributions', fontsize=16)

for i, feature in enumerate(features):
    ax = axes[i // 2][i % 2]
    sns.histplot(data=df, x=feature, hue='species', kde=True, palette='Set1', ax=ax, bins=15)
    ax.set_title(f'Distribution of {feature}')

plt.tight_layout()
plt.show()
```

```python
features = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
fig.suptitle('Box Plots: Outlier Detection', fontsize=16)

for i, feature in enumerate(features):
    ax = axes[i // 2][i % 2]
    sns.boxplot(data=df, x='species', y=feature, hue='species', palette='Set2', legend=False, ax=ax)
    ax.set_title(f'Box Plot: {feature}')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))
correlation = df[['sepal_length', 'sepal_width', 'petal_length', 'petal_width']].corr()
sns.heatmap(correlation, annot=True, fmt='.2f', cmap='coolwarm', linewidths=0.5, square=True)
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```
